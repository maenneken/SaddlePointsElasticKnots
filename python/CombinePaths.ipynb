```python
fileA = "../data/Paths/4_1/PathA.txt"
fileB = "../data/Paths/4_1/PathB-short.txt"
pathA = np.loadtxt(fileA, comments="#")
pathA = pathA[::-1]
pathB = np.loadtxt(fileB, comments="#")
print(pathA.shape)
print(pathB.shape)
```

```python
KnotA = pathA[-1,:75]

KnotB = pathB[0,:75]

```

```python
def shift_by_3(arr, k=1):
    # k = wie oft du um 3 verschieben willst
    return np.roll(arr, shift=3*k)
```

```python
def same_knot_full(a, b):
    if len(a) != len(b):
        return False
    
    step = 3
    n = len(a)
    
    # normal
    for k in range(n // step):
        if np.allclose(a, np.roll(b, 3*k)):
            return k
    
    # reversed (Punkte-Reihenfolge umdrehen)
    b_rev = b.reshape(-1, 3)[::-1].reshape(-1)
    
    for k in range(n // step):
        if np.allclose(a, np.roll(b_rev, 3*k)):
            return -k
    
    return False
```

```python
shiftfactor = same_knot_full(KnotA,KnotB)
print(shiftfactor)
```

```python
totalPath = [*pathA]  # convert initial array to list of rows

for K in pathB:
    if(shiftfactor < 0):
        #reversed order
        K[:75] = K[:75].reshape(-1, 3)[::-1].reshape(-1)
    K[:75] = shift_by_3(K[:75], abs(shiftfactor))
    totalPath.append(K)

totalPath = np.array(totalPath)
print(totalPath.shape)
```
